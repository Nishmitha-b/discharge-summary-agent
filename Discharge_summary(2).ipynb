```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "Qwen/Qwen2.5-7B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("Model Loaded Successfully")
```

```python
def ask_llm(prompt):

    messages = [
        {"role": "user", "content": prompt}
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        do_sample=False
    )

    response_ids = outputs[0][inputs.input_ids.shape[1]:]

    response = tokenizer.decode(
        response_ids,
        skip_special_tokens=True
    )

    return response.strip()
```

```python
MAX_STEPS = 20

def run_agent(full_text):

    memory = {
        "diagnoses": [],
        "medications": [],
        "pending_results": [],
        "allergies": [],
        "conflicts": [],
        "flags": []
    }

    trace = []

    step = 0

    while step < MAX_STEPS:

        step += 1

        trace.append(f"Step {step} started")

        if not memory["diagnoses"]:

            diagnosis = safe_extract(
                  extract_diagnosis,
                  full_text[:2000],
                   "Diagnosis Tool"
           )

            memory["diagnoses"].append(
                diagnosis
            )

            trace.append(
                "Diagnosis extraction completed"
            )

            continue

        if not memory["medications"]:

            meds = safe_extract(
    extract_medications,
    full_text[1000:4000],
    "Medication Tool"
)

            memory["medications"].append(
                meds
            )

            trace.append(
                "Medication extraction completed"
            )

            continue

        if not memory["pending_results"]:

            pending = safe_extract(
    extract_pending_results,
    full_text[:2500],
    "Pending Results Tool"
)

            memory["pending_results"].append(
                pending
            )

            trace.append(
                "Pending result extraction completed"
            )

            continue

        trace.append(
            "All required information collected"
        )

        break

    return memory, trace
```

```python
memory, trace = run_agent(full_text)

print(memory)

print()

for t in trace:
    print(t)
```

```python
def safe_extract(tool_func, text, tool_name):

    try:
        result = tool_func(text)

        if not result or result.strip() == "":
            raise ValueError("Empty response")

        return result

    except Exception as e:

        print(f"{tool_name} failed")

        return "MISSING"
```

```python
# ==========================================
# MEDICATION RECONCILIATION TOOL
# ==========================================

def reconcile_medications(admission_meds, discharge_meds):

    admission_set = set([m.strip().lower() for m in admission_meds])
    discharge_set = set([m.strip().lower() for m in discharge_meds])

    added = list(discharge_set - admission_set)
    stopped = list(admission_set - discharge_set)

    return {
        "added": added,
        "stopped": stopped,
        "needs_review": []
    }
```
